## ✊ 17. Rock, Paper, Scissors Game

### Goal:
    Decide the winner of a two-player game.

### Steps:

    Ask Player 1 and Player 2 for rock, paper or scissors.
    
    Rock beats scissors, scissors beats paper, paper beats rock.
    
    Use logical operators (and / or) to print the winner.
    
    If both choose the same → "Draw".

### Example:

    Player 1: rock
    Player 2: scissors
    Player 1 wins

```python
player1 = input("Player 1: ").lower()
player2 = input("Player 2: ").lower()
valid_moves = ("rock", "paper", "scissors")
if player1 not in valid_moves or player2 not in valid_moves:
    print("Invalid choice. Use rock, paper or scissors.")
elif player1 == player2:
    print("Draw")
elif (player1 == "rock" and player2 == "scissors") or \
     (player1 == "scissors" and player2 == "paper") or \
     (player1 == "paper" and player2 == "rock"):
    print("Player 1 wins")
else:
    print("Player 2 wins")
```

## 📝 25. Contact Form Validator

### Goal:
    Check which fields a user left empty on a sign-up form.

### Steps:

    Create form = {"name": "Sara", "email": "", "phone": "03001234567", "city": ""}.
    
    Loop through the form fields.
    
    If the value is truthy → print "<field> is filled".
    
    Otherwise → print "<field> is missing".
    
    Do not compare with == ""; use truthiness only.

### Example:

    name is filled
    email is missing
    phone is filled
    city is missing

```python
form = {"name": "Sara", "email": "", "phone": "03001234567", "city": ""}
for field in form:
    if form[field]:                  # non-empty string is truthy
        print(field, "is filled")
    else:                            # empty string is falsy
        print(field, "is missing")
```

    Remove blank answers from a customer survey.

    Create answers = ["Good", "", "Excellent", None, "Average", ""].

    Add an answer to a new list only if it is truthy.
    
    Print the cleaned list and how many answers were removed.

### Example:

    Cleaned: ['Good', 'Excellent', 'Average']
    Removed: 3

```python
answers = ["Good", "", "Excellent", None, "Average", ""]
cleaned_answers = []
for answer in answers:
    if answer:                       # "" and None are falsy, so they are skipped
        cleaned_answers.append(answer)
removed_count = len(answers) - len(cleaned_answers)
print("Cleaned:", cleaned_answers)
print("Removed:", removed_count)
```

## ⚙️ 30. App Settings Checker

### Goal:
    Check which settings a user has configured in an app.

### Steps:

    Create settings = {"theme": "dark", "volume": 0, "nickname": ""}.
    
    Loop through the keys ["theme", "volume", "nickname", "language"].
    
    Use settings.get(key) with a truthiness check.
    
    Truthy → "<key> is set to <value>", otherwise "<key> is empty or missing".
    
    Add a comment explaining why volume (0) is treated as empty.

### Example:

    theme is set to dark
    volume is empty or missing
    nickname is empty or missing
    language is empty or missing

```python
settings = {"theme": "dark", "volume": 0, "nickname": ""}
keys_to_check = ["theme", "volume", "nickname", "language"]
for key in keys_to_check:
    value = settings.get(key)        # get() returns None if the key is missing
    # volume = 0 is falsy in Python, so it is treated as "empty" just like
    # "" and None. (Real apps may need "is None" if 0 should count as set.)
    if value:
        print(key, "is set to", value)
    else:
        print(key, "is empty or missing")
```

## 🧱 47. Largest Square Tile

### Goal:
    A floor is 48 ft by 18 ft. Find the biggest square tile that fits without cutting.

### Steps:

    Ask for the length and the width.
    
    Use a while loop with Euclid's method: while b is not 0, set a, b = b, a % b.
    
    The answer is the final value of a (the GCD).
    
    Print the tile size.

### Example:

    Length: 48
    Width: 18
    Largest tile = 6 ft

```python
floor_length = int(input("Length: "))
floor_width = int(input("Width: "))
a = floor_length
b = floor_width
while b != 0:
    a, b = b, a % b          # the final value of a is the GCD
print("Largest tile =", a, "ft")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
temps = [30, 32, 28, 35, 31, 29, 36]
# First loop: add all temperatures
temperature_total = 0
for temperature in temps:
    temperature_total = temperature_total + temperature
average_temperature = temperature_total / len(temps)
# Second loop: count days above the average
days_above_average = 0
for temperature in temps:
    if temperature > average_temperature:
        days_above_average += 1
print("Average =", format(average_temperature, ".2f"))
print("Days above average =", days_above_average)
```
